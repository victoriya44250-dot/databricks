```python
from databricks.ai_search.client import AISearchClient

INDEX_NAME = (
    "workspace.viktoriia_kalenichenko_lab12.document_chunks_index"
)

client = AISearchClient()
index = client.get_index(index_name=INDEX_NAME)

# use the same list for various questions
excluded_pages = {3, 4, 5, 6, 96, 97}
allowed_pages = [page for page in range(3, 147)
    if page not in excluded_pages]

results = index.similarity_search(
    query_text="What agricultural risk management tools are used in Italy?",
    query_type="ANN",
    columns=["chunk_id", "page_number", "chunk_text"],
    filters={"page_number": allowed_pages},
    num_results=5,
)

column_names = [column["name"] for column in results["manifest"]["columns"]]

for rank, values in enumerate(
    results.get("result", {}).get("data_array", []),
    start=1):
    row = dict(zip(column_names, values))
    print(f"\n===== Result {rank}: page {row['page_number']} =====\n")
    print(row["chunk_text"])
```
