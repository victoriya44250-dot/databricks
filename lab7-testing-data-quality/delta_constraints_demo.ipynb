```python
%sql

CREATE TABLE IF NOT EXISTS
dbr_dev_ua5816bd.viktoriia_kalenichenko_silver.lab7_delta_constraints_demo (
    AREA STRING NOT NULL,
    ITEM STRING NOT NULL,
    YEAR INT NOT NULL,
    PRODUCTION_QUANTITY DOUBLE NOT NULL,
    AREA_HARVESTED DOUBLE NOT NULL
)
USING DELTA;
```

```python
%sql

-- 1. Year must be between 1990 and 2022
ALTER TABLE dbr_dev_ua5816bd.viktoriia_kalenichenko_silver.lab7_delta_constraints_demo
ADD CONSTRAINT valid_year
CHECK (YEAR BETWEEN 1990 AND 2022);

-- 2. Production cannot be negative
ALTER TABLE dbr_dev_ua5816bd.viktoriia_kalenichenko_silver.lab7_delta_constraints_demo
ADD CONSTRAINT valid_production
CHECK (PRODUCTION_QUANTITY >= 0);

-- 3. Harvested area cannot be negative
ALTER TABLE dbr_dev_ua5816bd.viktoriia_kalenichenko_silver.lab7_delta_constraints_demo
ADD CONSTRAINT valid_area
CHECK (AREA_HARVESTED >= 0);

```
